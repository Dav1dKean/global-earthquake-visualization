## 12. Investigate Missing Earthquake Depth Values

The initial data-quality check showed that some earthquake events are missing depth measurements.

Depth is important for one of the planned visualizations, so these missing values must be examined before deciding how to handle them.

Rather than replacing missing depths with an average value, the project will preserve the missing values and exclude those records only from visualizations that specifically require depth. This avoids introducing artificial depth measurements into the scientific data.

```python
# Count earthquakes that do not have a recorded depth.

missing_depth = df["depth_km"].isnull().sum()

# Calculate what percentage of the complete dataset is missing depth.

missing_depth_percent = (
    missing_depth / len(df)
) * 100

print("Earthquakes missing depth:", missing_depth)
print(
    "Percentage missing depth:",
    round(missing_depth_percent, 2),
    "%"
)
```

## 16. Define the Correct Order of Magnitude Categories

Magnitude classes represent increasing earthquake strength, so they should appear in scientific order rather than alphabetical order.

The categories are therefore defined as an ordered categorical variable:

**Light → Moderate → Strong → Major → Great**

This ensures that later charts display increasing earthquake strength from left to right.

```python
# Define the scientifically meaningful order
# of the earthquake magnitude categories.

magnitude_order = [
    "Light (M4.5-4.9)",
    "Moderate (M5-5.9)",
    "Strong (M6-6.9)",
    "Major (M7-7.9)",
    "Great (M8+)"
]

# Convert magnitude_class into an ordered categorical variable.

df["magnitude_class"] = pd.Categorical(
    df["magnitude_class"],
    categories=magnitude_order,
    ordered=True
)

# Verify the categories and their order.

df["magnitude_class"].value_counts().sort_index()
```

## Interactive Magnitude-Class Visualization

# View 1 — Where Do Earthquakes Occur?

The first visualization introduces the geographic distribution of earthquakes around the world.

The main question is:

**Where have magnitude 4.5 or greater earthquakes occurred globally?**

An interactive world map displays earthquake epicenters using latitude and longitude. Users can select a year and minimum earthquake magnitude to reduce visual clutter and explore geographic patterns.

Each point represents one recorded earthquake. Hovering over a point reveals the earthquake's location, date, magnitude, and depth.

This geographic view establishes the spatial context for the rest of the scientific story.

```python
# ---------------------------------------------------------
# VIEW 1: WHERE DO EARTHQUAKES OCCUR?
# Interactive Global Earthquake Map
# ---------------------------------------------------------

import ipywidgets as widgets
from IPython.display import display, clear_output
from google.colab import output

output.enable_custom_widget_manager()


# ---------------------------------------------------------
# Create the year selector.
# ---------------------------------------------------------

map_year_slider = widgets.IntSlider(
    value=2025,
    min=1900,
    max=2026,
    step=1,
    description="Year:",
    continuous_update=False,
    layout=widgets.Layout(width="700px")
)


# ---------------------------------------------------------
# Create the minimum magnitude slider.
# ---------------------------------------------------------

magnitude_slider = widgets.FloatSlider(
    value=4.5,
    min=4.5,
    max=8.0,
    step=0.5,
    description="Min Mag:",
    continuous_update=False,
    readout_format=".1f",
    layout=widgets.Layout(width="700px")
)


# ---------------------------------------------------------
# Create a Reset button.
# ---------------------------------------------------------

map_reset_button = widgets.Button(
    description="Reset"
)


# ---------------------------------------------------------
# Create an information label.
#
# This will show:
# - how many earthquakes are currently displayed
# - what marker size means
# - a warning when 2026 is selected
# ---------------------------------------------------------

map_info = widgets.HTML()


# ---------------------------------------------------------
# Create an output area for the map.
# ---------------------------------------------------------

map_output = widgets.Output()


# ---------------------------------------------------------
# Function that updates the map.
# ---------------------------------------------------------

def update_earthquake_map(change=None):

    # Read the currently selected controls.
    selected_year = map_year_slider.value
    minimum_magnitude = magnitude_slider.value

    # Filter the dataset to the selected year
    # and minimum magnitude.
    map_df = df[
        (df["year"] == selected_year) &
        (df["magnitude"] >= minimum_magnitude)
    ].copy()

    # Count how many earthquakes match the current filters.
    event_count = len(map_df)

    # -----------------------------------------------------
    # Update the information displayed above the map.
    # -----------------------------------------------------

    if selected_year == 2026:

        map_info.value = (
            f"<b>Events shown:</b> {event_count:,}"
            "<br>"
            "<b>Marker size:</b> larger circles represent stronger earthquakes."
            "<br>"
            "<span style='color:#9a6700;'>"
            "<b>Note:</b> 2026 contains data only through May 14, 2026."
            "</span>"
        )

    else:

        map_info.value = (
            f"<b>Events shown:</b> {event_count:,}"
            "<br>"
            "<b>Marker size:</b> larger circles represent stronger earthquakes."
        )


    # -----------------------------------------------------
    # Draw the updated map.
    # -----------------------------------------------------

    with map_output:

        clear_output(wait=True)

        fig = px.scatter_geo(
            map_df,

            # Geographic position.
            lat="lat",
            lon="lon",

            # Magnitude controls marker size.
            size="magnitude",

            # Limit maximum marker size so the map
            # remains readable.
            size_max=12,

            # Main hover heading.
            hover_name="location_desc",

            # Additional information shown on hover.
            hover_data={
                "date": True,
                "magnitude": ":.1f",
                "depth_km": ":.1f",
                "lat": False,
                "lon": False
            },

            title=(
                f"Global Earthquakes in {selected_year} "
                f"(Magnitude ≥ {minimum_magnitude})"
            ),

            projection="natural earth"
        )


        # -------------------------------------------------
        # Improve marker visibility without making
        # overlapping events too difficult to interpret.
        # -------------------------------------------------

        fig.update_traces(
            marker=dict(
                opacity=0.55,
                line=dict(width=0.3)
            )
        )


        # -------------------------------------------------
        # Improve geographic context.
        # -------------------------------------------------

        fig.update_geos(
            showcoastlines=True,
            showcountries=True,
            showland=True
        )


        # -------------------------------------------------
        # Improve layout.
        # -------------------------------------------------

        fig.update_layout(
            height=600,
            margin=dict(
                l=10,
                r=10,
                t=70,
                b=10
            )
        )

        fig.show()


# ---------------------------------------------------------
# Update the map when the year changes.
# ---------------------------------------------------------

map_year_slider.observe(
    update_earthquake_map,
    names="value"
)


# ---------------------------------------------------------
# Update the map when minimum magnitude changes.
# ---------------------------------------------------------

magnitude_slider.observe(
    update_earthquake_map,
    names="value"
)


# ---------------------------------------------------------
# Reset function.
# ---------------------------------------------------------

def reset_map(button):

    # Return to a complete recent year.
    map_year_slider.value = 2025
    magnitude_slider.value = 4.5


map_reset_button.on_click(reset_map)


# ---------------------------------------------------------
# Display the controls and visualization.
# ---------------------------------------------------------

display(map_year_slider)
display(magnitude_slider)
display(map_reset_button)

# Display information about the current map.
display(map_info)

# Display the map.
display(map_output)


# Draw the initial map.
update_earthquake_map()
```

# View 2 — How Has Earthquake Activity Changed Over Time?

This visualization examines how the number of recorded earthquakes changes across the historical dataset.

A line chart is used because time is ordered and the goal is to show patterns and changes across years.

Users can adjust the minimum magnitude to compare how the historical pattern changes when only stronger earthquakes are included.

Because earthquake monitoring technology and global reporting improved over time, increases in recorded earthquake counts should not automatically be interpreted as evidence that earthquakes have become physically more frequent.

```python
# ---------------------------------------------------------
# VIEW 2: HOW HAS EARTHQUAKE ACTIVITY CHANGED OVER TIME?
# Interactive Yearly Earthquake Timeline
# ---------------------------------------------------------

import numpy as np
import pandas as pd
import ipywidgets as widgets

from IPython.display import display, clear_output


# ---------------------------------------------------------
# Create a minimum magnitude slider.
# ---------------------------------------------------------

timeline_mag_slider = widgets.FloatSlider(
    value=4.5,
    min=4.5,
    max=8.0,
    step=0.5,
    description="Min Mag:",
    continuous_update=False,
    readout_format=".1f",
    layout=widgets.Layout(width="700px")
)


# ---------------------------------------------------------
# Create a Reset button.
# ---------------------------------------------------------

timeline_reset_button = widgets.Button(
    description="Reset"
)


# ---------------------------------------------------------
# Create an explanatory text area.
# ---------------------------------------------------------

timeline_info = widgets.HTML()


# ---------------------------------------------------------
# Create an output area for the chart.
# ---------------------------------------------------------

timeline_output = widgets.Output()


# ---------------------------------------------------------
# Function that updates the timeline.
# ---------------------------------------------------------

def update_timeline(change=None):

    # Read the selected minimum magnitude.
    minimum_magnitude = timeline_mag_slider.value

    # -----------------------------------------------------
    # Filter the dataset.
    #
    # 2026 is excluded because it is only a partial year.
    # -----------------------------------------------------

    filtered_df = df[
        (df["magnitude"] >= minimum_magnitude) &
        (df["year"] <= 2025)
    ].copy()


    # -----------------------------------------------------
    # Count earthquake records for each year.
    # -----------------------------------------------------

    yearly_counts = (
        filtered_df
        .groupby("year")
        .size()
        .reset_index(name="Number of Earthquakes")
    )


    # -----------------------------------------------------
    # Create a complete year sequence from 1900 to 2025.
    #
    # This makes missing years explicit instead of
    # silently skipping them.
    # -----------------------------------------------------

    all_years = pd.DataFrame({
        "year": range(1900, 2026)
    })

    yearly_counts = all_years.merge(
        yearly_counts,
        on="year",
        how="left"
    )


    # -----------------------------------------------------
    # Mark known incomplete / missing coverage years.
    #
    # 2005 has no records.
    # 2006 contains only Dec 8–31.
    #
    # Setting these values to NaN tells Plotly to leave
    # a visible break in the line.
    # -----------------------------------------------------

    yearly_counts.loc[
        yearly_counts["year"].isin([2005, 2006]),
        "Number of Earthquakes"
    ] = np.nan


    # -----------------------------------------------------
    # Update explanatory text shown above the chart.
    # -----------------------------------------------------

    timeline_info.value = (
        f"<b>Selected threshold:</b> Magnitude ≥ {minimum_magnitude:.1f}"
        "<br>"
        "<b>How to read this view:</b> "
        "Each point represents the number of recorded earthquakes "
        "during one year."
        "<br>"
        "<b>Important limitation:</b> Earlier historical counts may be "
        "affected by changes in earthquake monitoring and reporting."
        "<br>"
        "<b>Data coverage note:</b> "
        "2005 has no records in this dataset, and 2006 contains only "
        "partial coverage from December 8–31. "
        "The line is intentionally broken across this period."
        "<br>"
        "<b>Displayed period:</b> 1900–2025. "
        "The partial 2026 data are excluded from the yearly trend."
    )


    # -----------------------------------------------------
    # Draw the updated chart.
    # -----------------------------------------------------

    with timeline_output:

        clear_output(wait=True)

        fig = px.line(
            yearly_counts,
            x="year",
            y="Number of Earthquakes",
            markers=True,

            title=(
                f"Recorded Earthquakes by Year "
                f"(Magnitude ≥ {minimum_magnitude:.1f})"
            ),

            hover_data={
                "year": True,
                "Number of Earthquakes": ":,"
            }
        )


        # -------------------------------------------------
        # Keep the line broken across missing values.
        # -------------------------------------------------

        fig.update_traces(
            connectgaps=False,
            marker=dict(size=5),
            line=dict(width=2)
        )


        # -------------------------------------------------
        # Add an annotation identifying the coverage gap.
        # -------------------------------------------------

        fig.add_annotation(
            x=2005.5,

            # Place annotation near the bottom of the chart.
            y=0,

            text=(
                "Data coverage gap<br>"
                "2005: no records<br>"
                "2006: partial year"
            ),

            showarrow=True,

            # Point the arrow toward the missing section.
            ax=0,
            ay=-80,

            align="center"
        )


        # -------------------------------------------------
        # Improve chart readability.
        # -------------------------------------------------

        fig.update_layout(
            xaxis_title="Year",
            yaxis_title="Number of Recorded Earthquakes",
            template="plotly_white",
            hovermode="x unified",
            height=550,
            margin=dict(
                l=70,
                r=30,
                t=90,
                b=70
            )
        )


        # -------------------------------------------------
        # Keep the x-axis focused on the actual data range.
        # -------------------------------------------------

        fig.update_xaxes(
            dtick=10,
            range=[1900, 2025]
        )


        # -------------------------------------------------
        # Display the chart.
        # -------------------------------------------------

        fig.show()


# ---------------------------------------------------------
# Update the chart when the magnitude threshold changes.
# ---------------------------------------------------------

timeline_mag_slider.observe(
    update_timeline,
    names="value"
)


# ---------------------------------------------------------
# Reset function.
# ---------------------------------------------------------

def reset_timeline(button):

    timeline_mag_slider.value = 4.5


timeline_reset_button.on_click(reset_timeline)


# ---------------------------------------------------------
# Display the controls and visualization.
# ---------------------------------------------------------

display(timeline_mag_slider)
display(timeline_reset_button)
display(timeline_info)
display(timeline_output)


# ---------------------------------------------------------
# Draw the initial chart.
# ---------------------------------------------------------

update_timeline()
```
